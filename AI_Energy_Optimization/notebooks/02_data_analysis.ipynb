# Step 2: Exploratory Data Analysis (EDA)
Analyze energy consumption trends, correlations, and identify inefficiencies.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Data
energy = pd.read_csv('../data/energy_readings.csv')
weather = pd.read_csv('../data/weather_data.csv')
energy['timestamp'] = pd.to_datetime(energy['timestamp'])
weather['timestamp'] = pd.to_datetime(weather['timestamp'])

df = pd.merge(energy, weather, on='timestamp')
print(df.head())
```

```python
# Plot Daily Trend
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='timestamp', y='energy_kwh', hue='building_id')
plt.title('Energy Consumption over Time')
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df[['energy_kwh', 'temperature', 'humidity']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```
