# Step 5: Optimization Logic
Apply rule-based logic to recommend HVAC adjustments.

```python
import pandas as pd
import numpy as np

# Sample data to initialize the DataFrame
data = {
    'occupancy_level': ['Low', 'Low', 'High', 'Low'],
    'temperature': [26, 22, 28, 27]
}
df = pd.DataFrame(data)

# Refactored using vectorized operations for performance and to handle potential edge cases
conditions = [
    (df['occupancy_level'] == 'Low') & (df['temperature'] > 25),
    (df['occupancy_level'] == 'Low') & (df['temperature'] <= 25)
]
choices = ['Turn off AC', 'Set AC to 26C']

df['action'] = np.select(conditions, choices, default='No Action')
df
```
